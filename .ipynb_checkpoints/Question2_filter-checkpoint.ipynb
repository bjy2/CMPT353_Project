```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt
from statsmodels.nonparametric.smoothers_lowess import lowess
import os

# --- Kalman Filter ---
def simple_kalman_filter(z, R=0.01, Q=0.1):
    n = len(z)
    xhat = np.zeros(n)
    P = np.zeros(n)
    xhat[0] = z[0]
    P[0] = 1.0
    for k in range(1, n):
        xhat_minus = xhat[k-1]
        P_minus = P[k-1] + Q
        K = P_minus / (P_minus + R)
        xhat[k] = xhat_minus + K * (z[k] - xhat_minus)
        P[k] = (1 - K) * P_minus
    return xhat

# --- Butterworth Filter ---
def butter_lowpass_filter(data, cutoff, fs, order=4):
    nyq = 0.5 * fs
    normal_cutoff = cutoff / nyq
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    return filtfilt(b, a, data)

# --- Mask low variation area ---
def mask_low_variation(signal, threshold=0.0001):
    grad = np.abs(np.gradient(signal))
    mask = grad > threshold
    masked_signal = signal.copy()
    masked_signal[~mask] = np.nan
    return masked_signal

# --- Load the CSV (run1-ankle.csv) ---
file_path = os.path.join("data", "run1-ankle.csv")
df = pd.read_csv(file_path)

# --- Compute total acceleration and sampling frequency ---
df['acc_total'] = np.sqrt(df['gFx']**2 + df['gFy']**2 + df['gFz']**2)

# --- Get data from 5s to 25s ---
start_time = df['time'].iloc[0] + 5
end_time = start_time + 20 
df = df[(df['time'] >= start_time) & (df['time'] <= end_time)].reset_index(drop=True)

# --- Recalculate sampling rate ---
time_diff = np.diff(df['time'])
fs = 1 / np.mean(time_diff) if len(time_diff) > 0 else 100

# --- Apply filters ---
df['butter'] = butter_lowpass_filter(df['acc_total'], cutoff=5, fs=fs)
df['loess'] = lowess(df['acc_total'], df['time'], frac=0.001, return_sorted=False)
df['kalman'] = simple_kalman_filter(df['acc_total'].values)

# --- Apply cleaning ---
df['butter_clean'] = mask_low_variation(df['butter'])
df['loess_clean'] = mask_low_variation(df['loess'])
df['kalman_clean'] = mask_low_variation(df['kalman'])

# --- Plot Raw Acceleration Only ---
plt.figure(figsize=(12, 5))
plt.plot(df['time'], df['acc_total'], label='Raw', color='black')
plt.title("Raw Acceleration - run1-ankle (Middle Segment)")
plt.xlabel("Time (s)")
plt.ylabel("Total Acceleration")
plt.grid(True)
plt.legend()
plt.savefig("plots/raw_run1-ankle_segment.png")
plt.show()

# --- Plot Butterworth (cleaned) ---
plt.figure(figsize=(12, 5))
plt.plot(df['time'], df['butter_clean'], label='Butterworth (cleaned)', color='blue')
plt.title("Butterworth Filter (Masked) - run1-ankle (Segment)")
plt.xlabel("Time (s)")
plt.ylabel("Filtered Acceleration")
plt.grid(True)
plt.legend()
plt.savefig("plots/butterworth_run1-ankle_masked_segment.png")
plt.show()

# --- Plot Loess (cleaned) ---
plt.figure(figsize=(12, 5))
plt.plot(df['time'], df['loess_clean'], label='Loess (cleaned)', color='green')
plt.title("Loess Filter (Masked) - run1-ankle (Segment)")
plt.xlabel("Time (s)")
plt.ylabel("Filtered Acceleration")
plt.grid(True)
plt.legend()
plt.savefig("plots/loess_run1-ankle_masked_segment.png")
plt.show()

# --- Plot Kalman (cleaned) ---
plt.figure(figsize=(12, 5))
plt.plot(df['time'], df['kalman_clean'], label='Kalman (cleaned)', color='red')
plt.title("Kalman Filter (Masked) - run1-ankle (Segment)")
plt.xlabel("Time (s)")
plt.ylabel("Filtered Acceleration")
plt.grid(True)
plt.legend()
plt.savefig("plots/kalman_run1-ankle_masked_segment.png")
plt.show()
```
